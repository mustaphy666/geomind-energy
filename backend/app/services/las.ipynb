```python
import numpy as np

# Set seed for reproducibility
np.random.seed(42)

# Generate synthetic depth array (e.g., 2000ft to 2200ft at 0.5ft interval)
depth = np.arange(2000.0, 2200.5, 0.5)
n_samples = len(depth)

# Generate realistic synthetic well log curves
# Gamma Ray (GR) in API: range ~20 - 140
gr = 60 + 30 * np.sin(depth / 15) + np.random.normal(0, 5, n_samples)
gr = np.clip(gr, 15, 150)

# Bulk Density (RHOB) in g/cm3: range ~2.1 - 2.65
rhob = 2.45 - 0.2 * np.sin(depth / 15) + np.random.normal(0, 0.03, n_samples)
rhob = np.clip(rhob, 2.05, 2.71)

# Neutron Porosity (NPHI) in v/v or %: range ~0.1 - 0.35
nphi = 0.15 + 0.12 * np.sin(depth / 15) + np.random.normal(0, 0.015, n_samples)
nphi = np.clip(nphi, 0.05, 0.45)

# Deep Resistivity (RT) in ohm.m: range ~1 - 100
rt = 10 ** (1.0 + 0.8 * np.sin(depth / 20) + np.random.normal(0, 0.1, n_samples))
rt = np.clip(rt, 0.2, 200)

# Sonic Interval Transit Time (DT) in us/ft: range ~60 - 120
dt = 75 + 25 * np.sin(depth / 15) + np.random.normal(0, 2, n_samples)
dt = np.clip(dt, 50, 140)

# Caliper (CALI) in inches: bit size ~8.5 in, with occasional washouts
cali = 8.5 + np.maximum(0, np.random.normal(0.1, 0.3, n_samples))

# Build LAS file text strictly following LAS 2.0 standard
las_content = f"""~VERSION INFORMATION
 VERS.                  2.0: CWLS LOG ASCII STANDARD - VERSION 2.0
 WRAP.                   NO: ONE LINE PER DEPTH STEP
~WELL INFORMATION BLOCK
#MNEM.UNIT       DATA                           DESCRIPTION
#----.----       ------------------------------ ----------------------------------
 STRT.FT         2000.0000                      :START DEPTH
 STOP.FT         2200.0000                      :STOP DEPTH
 STEP.FT            0.5000                      :STEP
 NULL.           -999.2500                      :NULL VALUE
 COMP.           GEOMIND AI                     :COMPANY
 WELL.           WELL-01                        :WELL
 FLD .           EXPLORATION FIELD              :FIELD
 LOC .           OFFSHORE                       :LOCATION
 SRVC.           GEOPHYSICAL SERVICES           :SERVICE COMPANY
 DATE.           2026-08-20                     :LOG DATE
 UWI .           12-345-67890-00                :UNIQUE WELL ID
~CURVE INFORMATION BLOCK
#MNEM.UNIT       API CODE                       CURVE DESCRIPTION
#----.----       ------------------------------ ----------------------------------
 DEPTH.FT                                       :1  DEPTH
 GR   .GAPI                                     :2  GAMMA RAY
 RHOB .G/CC                                     :3  BULK DENSITY
 NPHI .V/V                                      :4  NEUTRON POROSITY
 RT   .OHMM                                     :5  TRUE RESISTIVITY
 DT   .US/F                                     :6  SONIC TRANSIT TIME
 CALI .IN                                       :7  CALIPER
~PARAMETER INFORMATION BLOCK
#MNEM.UNIT       VALUE                          DESCRIPTION
#----.----       ------------------------------ ----------------------------------
 BSIZ.IN            8.5000                      :BIT SIZE
~A DEPTH       GR         RHOB       NPHI       RT         DT         CALI
"""

# Append data lines
for d, g, r, n, res, t, c in zip(depth, gr, rhob, nphi, rt, dt, cali):
    las_content += f"{d:10.4f} {g:10.4f} {r:10.4f} {n:10.4f} {res:10.4f} {t:10.4f} {c:10.4f}\n"

output_filename = "synthetic_well_log.las"
with open(output_filename, "w") as f:
    f.write(las_content)

print(f"File saved successfully as {output_filename}")
```
